# Unicode Encodings

(a) What are some reasons to prefer training our tokenizer on UTF-8 encoded bytes, rather than 
UTF-16 or UTF-32?

```python
len("hello, world".encode('utf-8'))
```

```python
len("hello, world".encode('utf-16'))
```

```python
len("hello, world".encode('utf-32'))
```

In general, memory savings and less complexity are both good arguments to train on UTF-8 over other protocols.

- UTF-8 is much more space/memory efficient (~2x) for English text than other encodings (UTF-16, UTF-32), and English usually consists of majority of training data.

- Majority of internet is using UTF-8 (As of January 2026 , UTF-8 is used by 98.9% of surveyed web sites).
Training on UTF-16/UTF-32 would require to convert most of these webpages, which might introduce tricky data bugs. Additionally UTF-16/UTF-32 are more complex protocols than UTF-8 (e.g. byte order needs to be specified).

(b) Consider the following (incorrect) function, which is intended to decode a UTF-8 byte string 
into a Unicode string. Why is this function incorrect? Provide an example of an input byte 
string that yields incorrect results.

```python
def decode_utf8_bytes_to_str_wrong(bytestring: bytes):
    return "".join([bytes([b]).decode("utf-8") for b in bytestring])
```

```python
decode_utf8_bytes_to_str_wrong("hello".encode("utf-8"))
```

```python
decode_utf8_bytes_to_str_wrong("おはよう".encode("utf-8"))
```

Decofding code expects each character to have 1-byte, so if we try to decode characters that have multiple bytes in UTF-8 (like japanese hiragana), the buggy code throws UnicodeDecodeError.

(c) Give a two-byte sequence that does not decode to any Unicode character(s).

```python
twice_11000001 = b"\xC1\xC1"
```

```python
twice_11000001.decode('utf-8')
```

In 2-byte sequence, UTF-8 expects "10" always as leading bits in the 2nd byte. So if two bytes both have "11" as leading bits, it results in an invalid UTF-8 character.
